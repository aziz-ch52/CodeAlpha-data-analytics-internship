# 01. Data Preparation for Visualization
**Objective:** Load the raw global ecommerce sales data, clean it, and export a finalized dataset specifically formatted for plotting.

*We maintain the read-only rule for the raw data here.*


```python
import pandas as pd
import numpy as np
import os

# Ensure output directories exist
os.makedirs('../data/processed', exist_ok=True)

# 1. Load the RAW data
df = pd.read_csv('../data/raw/global_ecommerce_sales.csv')

# 2. Basic Cleaning (Standardizing names, dropping exact dupes)
df.columns = df.columns.str.lower().str.replace(' ', '_')
df = df.drop_duplicates()

# Fill numerical nulls with median, categorical with 'Unknown'
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna('Unknown')

# Ensure date columns are datetime (adjust 'order_date' to your actual date column name)
if 'order_date' in df.columns:
    df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

# 3. Export for Visualization Pipeline
df.to_csv('../data/processed/cleaned_ecommerce_sales.csv', index=False)
print("Data prepared and exported to ../data/processed/cleaned_ecommerce_sales.csv")
```
