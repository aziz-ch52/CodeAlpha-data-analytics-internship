# 04. Bivariate Analysis
**Objective:** Pit two variables against each other to find linear trends or logical separations.

*Working entirely off the processed data.*


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/ecommerce_cleaned.csv')
sns.set_theme(style="darkgrid")

# Numeric vs Numeric
if 'price' in df.columns and 'quantity' in df.columns:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x='price', y='quantity', alpha=0.5)
    plt.title('Price vs Quantity Relationship')
    plt.show()

# Categorical vs Numeric
if 'category' in df.columns and 'price' in df.columns:
    plt.figure(figsize=(12, 6))
    sns.boxplot(data=df, x='price', y='category')
    plt.title('Price Distribution Across Categories')
    plt.show()
```
