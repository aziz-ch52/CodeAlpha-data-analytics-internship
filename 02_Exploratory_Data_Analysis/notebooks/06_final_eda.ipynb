# 06. Final EDA
**Objective:** Synthesize findings into actionable business metrics and engineer final features.

*This is what gets presented or fed into a model.*


```python
import pandas as pd

df = pd.read_csv('../data/processed/ecommerce_cleaned.csv')

# Feature Engineering
if 'price' in df.columns and 'quantity' in df.columns:
    df['total_revenue'] = df['price'] * df['quantity']

# High-Level Aggregations
print("--- Executive Summary ---")
print(f"Total Transactions: {len(df)}")
if 'total_revenue' in df.columns:
    print(f"Total Revenue Generated: ${df['total_revenue'].sum():,.2f}")

# Top Performing Categories
if 'category' in df.columns and 'total_revenue' in df.columns:
    category_performance = df.groupby('category')['total_revenue'].sum().sort_values(ascending=False)
    print("\nRevenue by Category:")
    display(category_performance.head())

# Final Export for Modeling/Dashboarding
# df.to_csv('../data/processed/ecommerce_final_features.csv', index=False)
```
