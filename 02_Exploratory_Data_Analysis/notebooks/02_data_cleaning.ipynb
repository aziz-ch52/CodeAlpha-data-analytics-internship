# 02. Data Cleaning
**Objective:** Permanently fix structural issues identified in Notebook 01 and export the single source of truth for the rest of your analysis.

*Never perform analytical calculations here.*


```python
import pandas as pd

# Load raw data
df = pd.read_csv('../data/raw/ecommerce_retail_transactions_synthetic.csv')

# Strip exact duplicates
df = df.drop_duplicates()

# Standardize column names
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Handle Missing Values
# Adjust these columns based on the actual raw data columns
if 'customer_id' in df.columns:
    df = df.dropna(subset=['customer_id'])
    
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna('Unknown')

# Correct Data Types
if 'transaction_date' in df.columns:
    df['transaction_date'] = pd.to_datetime(df['transaction_date'], errors='coerce')

# Export to Processed Folder
# THIS BECOMES THE NEW SOURCE OF TRUTH
df.to_csv('../data/processed/ecommerce_cleaned.csv', index=False)
print("Cleaned data exported successfully to data/processed/")
```
