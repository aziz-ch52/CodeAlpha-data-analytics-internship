# 01. Data Understanding
**Objective:** Load the raw data and expose its flaws. Do not clean anything yet.

*This notebook loads the data exactly as it is to understand the structure, types, and the extent of the mess.*


```python
import pandas as pd
import numpy as np


df_raw = pd.read_csv('../data/raw/ecommerce_retail_transactions_synthetic.csv')

# Inspect shape and data types
print("Dataset Shape:", df_raw.shape)
display(df_raw.info())

#Quick statistical summary of numerical columns
display(df_raw.describe())

#Expose missing values
missing_data = df_raw.isnull().sum()
print("\nMissing Values per Column:")
print(missing_data[missing_data > 0])

#Check for exact duplicate rows
print(f"\nDuplicate Rows: {df_raw.duplicated().sum()}")
```
