# 03. Univariate Analysis
**Objective:** Analyze the distribution of individual variables and spot extreme outliers on the PROCESSED data.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load CLEANED data ONLY
df = pd.read_csv('../data/processed/ecommerce_cleaned.csv')
sns.set_theme(style="darkgrid")

# Numerical Distribution
if 'price' in df.columns:
    plt.figure(figsize=(10, 5))
    sns.histplot(df['price'], bins=50, kde=True)
    plt.title('Distribution of Price')
    plt.show()

# Boxplot for Outliers
if 'price' in df.columns:
    plt.figure(figsize=(10, 3))
    sns.boxplot(x=df['price'])
    plt.title('Outlier Detection: Price')
    plt.show()

# Categorical Frequency
if 'category' in df.columns:
    plt.figure(figsize=(12, 5))
    sns.countplot(y=df['category'], order=df['category'].value_counts().index)
    plt.title('Transaction Count by Category')
    plt.show()
```
