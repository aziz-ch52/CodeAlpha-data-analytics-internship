# Task 1: Web Scraping & Data Extraction (`books.toscrape.com`)

**Objective:** Extract, clean, validate, and export the complete 1,000-book catalog from [Books to Scrape](https://books.toscrape.com/) across all 50 pagination pages.

### Pipeline Workflow
1. **Environment Setup:** Configure system paths to import modular `src` components.
2. **Scraping:** Crawl all catalog pages using `BookScraper` with retry handling and polite delays.
3. **Data Cleaning:** Normalize GBP prices, map star ratings (`One`–`Five`) to integers, and deduplicate records.
4. **Data Export:** Persist raw snapshots to `data/raw/` and cleaned datasets to `data/processed/`.
5. **Verification:** Audit schema, missing values, and summary statistics.

## 1. Environment & Module Imports
Add the `01_Web_Scraping/` project root to `sys.path` so the modular pipeline scripts inside `src/` can be imported cleanly.

```python
import sys
from pathlib import Path

# Resolve 01_Web_Scraping/ root directory
project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.scraper import BookScraper
from src.data_cleaning import clean_books_data
from src.data_export import export_data

print(f"Project root loaded: {project_root}")
```

## 2. Scrape Catalog Pages
Initialize `BookScraper` and extract raw HTML records across all 50 pages.

```python
scraper = BookScraper(delay=0.2)
raw_books = scraper.scrape_all()

print(f"Total raw records scraped: {len(raw_books)}")
raw_books[:2]
```

## 3. Clean & Export Datasets
Transform raw records into a typed `pandas.DataFrame` and export both raw and processed files to `data/raw/` and `data/processed/`.

```python
df_books = clean_books_data(raw_books)
export_data(raw_books, df_books, base_dir=project_root)

print("Export completed successfully.")
df_books.head(10)
```

## 4. Data Quality & Summary Verification
Verify column data types, non-null counts, duplicate checks, and price/rating distributions.

```python
print("--- Dataset Info ---")
df_books.info()

print("\n--- Missing Values ---")
print(df_books.isnull().sum())

print(f"\n--- Duplicate URLs: {df_books['product_url'].duplicated().sum()} ---")

print("\n--- Summary Statistics ---")
df_books.describe()
```
